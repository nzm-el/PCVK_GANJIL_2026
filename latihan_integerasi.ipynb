```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = [
    "https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
    "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
    "https://picsum.photos/800/571" # Menggunakan URL alternatif
]
# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
    image = io.imread(url)                               #read image
    image = cv.resize(image, (0,0), fx=0.5, fy=0.5)      #resize image to half size
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)       #convert color to RGB
    final_frame = cv.hconcat((image, image_2))           #concatenate image
    cv2_imshow(final_frame)                              #show image
    print('\n')
```

```python
import cv2 as cv
from skimage import io
from google.colab.patches import cv2_imshow
import numpy as np

# 1. Membaca dan menyiapkan citra awal
url = "https://picsum.photos/800/571"
image = io.imread(url)
image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
image_asli = cv.cvtColor(image, cv.COLOR_BGR2RGB)

tinggi = image_asli.shape[0]
lebar = image_asli.shape[1]

# -------------------------------------------------------------
# 2A. Membuat Garis Vertikal (Manipulasi Array Piksel)
# -------------------------------------------------------------
img_vertikal = image_asli.copy()

# Looping pada sumbu y (tinggi) di posisi kolom tengah
for y in range(tinggi):
    img_vertikal[y, int(lebar/2)] = [255, 255, 255]

# Menggabungkan gambar asli dan hasil vertikal secara horizontal
hasil_vertikal = cv.hconcat((image_asli, img_vertikal))
print("Hasil Garis Vertikal:")
cv2_imshow(hasil_vertikal)


# -------------------------------------------------------------
# 2B. Membuat Garis Menyilang Diagonal (Menggunakan cv.line)
# -------------------------------------------------------------
img_diagonal = image_asli.copy()

# Menggambar garis dari sudut ke sudut
# Format cv.line: (image, titik_awal(x,y), titik_akhir(x,y), warna, ketebalan)
cv.line(img_diagonal, (0, 0), (lebar, tinggi), (255, 255, 255), 1)          # Kiri atas ke kanan bawah
cv.line(img_diagonal, (0, tinggi), (lebar, 0), (255, 255, 255), 1)          # Kiri bawah ke kanan atas

# Menggabungkan gambar asli dan hasil diagonal secara horizontal
hasil_diagonal = cv.hconcat((image_asli, img_diagonal))
print("\nHasil Garis Menyilang Diagonal:")
cv2_imshow(hasil_diagonal)
```

```python
import cv2 as cv
from skimage import io
from google.colab.patches import cv2_imshow
import numpy as np

# Membaca dan menyiapkan citra awal
url = "https://picsum.photos/800/571"
image = io.imread(url)
image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
image_asli = cv.cvtColor(image, cv.COLOR_BGR2RGB)

tinggi = image_asli.shape[0]
lebar = image_asli.shape[1]

# Membuat salinan gambar untuk dimanipulasi
img_garis_pendek = image_asli.copy()

# Menentukan posisi Y (tengah) dan batas X (panjang garis)
y_tengah = int(tinggi / 2)
x_mulai = int(lebar * 0.3)   # Garis dimulai pada titik 30% dari lebar gambar
x_selesai = int(lebar * 0.7) # Garis berakhir pada titik 70% dari lebar gambar

# Memanipulasi piksel array untuk membuat garis putih
for x in range(x_mulai, x_selesai):
    img_garis_pendek[y_tengah, x] = [255, 255, 255]

# Menggabungkan gambar asli dan gambar hasil secara horizontal
hasil = cv.hconcat((image_asli, img_garis_pendek))
cv2_imshow(hasil)
```

```python
import cv2 as cv
from skimage import io
from google.colab.patches import cv2_imshow
import numpy as np

# Membaca dan menyiapkan citra awal
url = "https://picsum.photos/800/571"
image = io.imread(url)
image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
image_asli = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Membuat salinan gambar untuk dimanipulasi
img_kotak = image_asli.copy()

# Menentukan batas koordinat untuk kotak putih (y1:y2, x1:x2)
# Nilai ini dapat disesuaikan untuk menggeser posisi atau mengubah ukuran kotak
y1, y2 = 80, 180
x1, x2 = 120, 220

# Memanipulasi sekumpulan piksel (slicing matrix) menjadi warna putih
img_kotak[y1:y2, x1:x2] = [255, 255, 255]

# Menggabungkan gambar asli dan gambar hasil secara horizontal
hasil = cv.hconcat((image_asli, img_kotak))
cv2_imshow(hasil)
```
