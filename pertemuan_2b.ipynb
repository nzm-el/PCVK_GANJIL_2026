**6. Membuat bentuk Geometri 2D dari OpenCV. Diawali dengan pembuatan
black image dengan tipe data int16.**

```python
import numpy as np

black_img = np.zeros(shape=(512, 512, 3), dtype=np.int16)
plt.imshow(black_img)
plt.show()
```


**Kemudian menambahkan bentuk persegi panjang sesuai koordinat pt1 dan pt2**

```python
# perhatikan koordinat titik2 pt1 dan pt2
cv2.rectangle(
    black_img, pt1=(384, 0), pt2=(510, 150), color=(0, 255, 0), thickness=10
)
plt.imshow(black_img)
plt.show()
```

**Selanjutnya ditambah menambahkan bentuk persegi sesuai koordinat pt1 dan
pt2 yang tertulis pada kode program**

```python
cv2.rectangle(
    black_img, pt1=(200, 200), pt2=(300, 300), color=(0, 0, 255), thickness=15
)
plt.imshow(black_img)
plt.show()
```

**Tahap selanjutnya ditambah menambahkan bentuk lingkaran sesuai radius
yang tertulis pada kode program.**

```python
cv2.circle(
    black_img, center=(100, 100), radius=50, color=(255, 0, 0), thickness=8
)
plt.imshow(black_img)
plt.show()
```

**Kemudian dilakukan penambahan garis sesuai koordinat pt1 dan pt2 sebagai
berikut**

```python
cv2.line(
    black_img, pt1=(0, 0), pt2=(512, 512), color=(255, 255, 255), thickness=5
)
plt.imshow(black_img)
plt.show()
```

**Penambahan text dengan font yang telah tertulis dengan ukuran yang sudah
ditentukan.**

```python
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(
    black_img,
    text="Hello",
    org=(10, 500),
    fontFace=font,
    fontScale=4,
    color=(255, 255, 0),
    thickness=2,
    lineType=cv2.LINE_AA,
)
plt.imshow(black_img)
plt.show()
```

**Pembuatan black image kembali dilakukan dengan tipe data int32**

```python
black_img2 = np.zeros(shape=(512, 512, 3), dtype=np.int32)
plt.imshow(black_img2)
```

**Berikut adalah kode program untuk inisialisasi NumPy array dengan tipe data
int32**

```python
vertices = np.array([[100, 300], [200, 200], [400, 300], [200, 400]], dtype=np.int32)
vertices
```

**Array tersebut kemudian di reshape sebagai berikut**

```python
pts = vertices.reshape(
    (-1, 1, 2)
)  # nilai 2 untuk menunjukkan bahwa tiap titik dibuat 3 channel yg mewakili R, G, dan B
pts
```

**Penambahan polyline pada black image kedua yang telah dibuat.**

```python
cv2.polylines(
    black_img2, [pts], isClosed=True, color=(255, 0, 0), thickness=5
)
plt.imshow(black_img2)
plt.show()
```

**PERTANYAAN PRAKTIKUM D3**

1. Perbedaan gambar yang ditampilkan tanpa dan dengan Matplotlib:

= Tanpa Matplotlib (cv2_imshow): Menampilkan gambar menggunakan format warna asli OpenCV yaitu BGR, sehingga warna asli bisa tampak tertukar (misal merah menjadi biru) jika belum dikonversi.

Dengan Matplotlib (plt.imshow): Menggunakan standar format warna RGB, menampilkan sumbu koordinat piksel di sekeliling gambar, serta mendukung pengaturan ukuran canvas (figsize) dan tata letak subplot.

2. Perbedaan dan pengaruh pembuatan black image antara tipe data int16 dan int32:

= int16: Mengalokasikan memori sebesar 16-bit per channel piksel dengan batas rentang angka yang lebih kecil.

int32: Mengalokasikan memori 32-bit yang lebih besar dan biasa digunakan agar kompatibel dengan fungsi geometri kompleks atau array vertices (seperti pada pembuatan poligon) guna menghindari risiko overflow.

3. Kegunaan from google.colab.patches import cv2_imshow:

= Digunakan untuk menampilkan gambar secara aman di lingkungan Google Colab, menggantikan fungsi bawaan OpenCV (cv2.imshow()) yang tidak dapat berjalan di browser dan dapat menyebabkan kernel session mengalami crash.
